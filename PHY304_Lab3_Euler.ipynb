```python

import numpy as np
import matplotlib.pyplot as plt
import os

def euler_newton(force, times, x0, v0, m, dt):
    x = [x0]
    v = [v0]
    for i in range (len(times)-1):
        a = force(times[i], x[i], v[i]) / m
        v.append( v[i] + a*dt)
        x.append( x[i] + v[i]*dt)
    return x, v
start = 0
stop = 10
nsteps = 100000
times,dt = np.linspace(start, stop, nsteps,retstep=True)
m=10
g=9.8

def gravity(t, x0, v0):
    output = -m*g
    return output

def f(x0, v0, g, time):
    ouput = x0 + v0*times - 1/2*g*t**2
    return output

print(gravity(0, x0, v0))

x0 = 10 
v0 = 0
```

```python
x, v = euler_newton(gravity,times,x0,v0,m,dt)
def quad(times,x0, v0):
    output= x0 + v0*times - 1/2*g*times**2
    return output

plt.plot(times,x)
plt.plot(times,quad(times,x0, v0),ls="dashed")
plt.xlabel("x (time)")
plt.ylabel("y (velocity)")
plt.show()

plt.plot(times,v)
plt.plot(times,v0-g*times,ls="dashed")
plt.xlabel("x (time)")
plt.ylabel("y (/position)")
plt.show()
```
