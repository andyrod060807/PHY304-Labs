## 2. Variables, assignment, and types

A **variable** gives a name to a value. In Python,

```python

```

````python
name = "Dr. Halliday"
bigint = 1000
flag = True
nflag = not flag
````

Because each value above was assigned to a variable, Jupyter did not automatically display anything. We can explicitly inspect a value or its type:


````python
type(name)
````

````python
nametype = type(name)
print(nametype)
````

## 3. Writing your own functions

Python includes many functions such as `print(...)` and `type(...)`, and packages provide many more such as `np.linspace(...)` and `plt.plot(...)`. We can also define our own.

The function below implements a line,

\[
f(x)=mx+b.
\]

Important details:

- `def` begins a function definition.
- The names in parentheses are the function's **parameters**.
- The colon `:` begins the function body.
- The indented lines belong to the function. **Indentation is part of Python syntax.**
- `return` sends a value back to the code that called the function.


````python
def f(x, m, b):
    output = m*x + b
    return output
````

Defining a function does not evaluate it. We evaluate it later by **calling** it with specific arguments:


````python
m = 1
b = 0
print(f(0, m, b))
````

## 4. Lists, NumPy arrays, and array arithmetic

Python has a built-in sequence type called a **list**. NumPy provides a different sequence type called an **array**. They look similar when printed, but arithmetic behaves differently.


````python
# A built-in Python list
data_py = [1, 2, 3, 4]

# A NumPy array created by calling np.array(...)
data_np = np.array([1, 2, 3, 4])
````

Compare what multiplication means for the two types:


````python
print("data_py:")
print(data_py)
print("4*data_py:")
print(4*data_py)

print("data_np:")
print(data_np)
print("4*data_np:")
print(4*data_np)
````

Multiplying a Python list by an integer repeats the list. Multiplying a NumPy array by a number performs element-by-element numerical multiplication. Both behaviors can be useful, but for numerical physics work we will usually want NumPy arrays.

## 5. Building arrays for plotting

NumPy has convenient functions for generating arrays:

- `np.ones(n)` — `n` entries equal to 1
- `np.zeros(n)` — `n` entries equal to 0
- `np.linspace(start, stop, number_of_points)` — evenly spaced values including both endpoints
- `np.arange(start, stop, step)` — values separated by `step`, usually stopping before `stop`

Here we create 20 x-values and evaluate our function `f` on the entire array.


````python
xs = np.linspace(-5, 5, 20)
ys = f(xs, m, b)
print(xs)
print(ys)
````

Notice that the same function `f` worked when `x` was a single number and when `x` was an entire NumPy array. The expression `m*x + b` is evaluated element-by-element on the array without us writing an explicit Python loop.

That is an example of **vectorized array arithmetic**. Vectorization is one of the main reasons NumPy code is concise and fast.

A related but distinct idea is **broadcasting**: NumPy has rules that allow arrays of compatible shapes—and scalars such as `m` and `b`—to participate in the same operation. Here the scalars are broadcast across every element of `xs`.


## 6. Plotting with Matplotlib

4. optionally add a grid or other formatting;
5. call `plt.tight_layout()`;
6. call `plt.show()`.

As physicists, **label axes and include units whenever the quantities have physical units.** If a quantity is intentionally unitless or only its relative scale matters, `a.u.` (“arbitrary units”) can be appropriate.


````python
plt.plot(xs, ys)
plt.xlabel("x (a.u.)")
plt.ylabel("y (a.u.)")
plt.tight_layout()
plt.show()
````

## 7. Flow control: `if`, `else`, `for`, `range`, `enumerate`, and `while`

**Flow control** determines which lines of code run and how many times they run.

### `for` and `range`
A `for` loop repeats a block once for each item in a sequence. `range(5)` produces the integers 0, 1, 2, 3, 4 for iteration.


````python
for i in range(6):
    print(i)

    if i == 5:
        print("i has reached 5!")
````

### `if` and `else`
`if` runs a block when a condition is `True`; `else` supplies the alternative.


````python
temperature = 18.0

if temperature >= 20.0:
    print("At or above threshold")
else:
    print("Below threshold")
````

### Looping directly over an array
You do not need an index if you only care about each value:


````python
measurements = np.array([0.8, 1.1, 1.4, 0.9])

for value in measurements:
    print(value)
````

### `enumerate`
Use `enumerate(...)` when you want both the index and the value:


````python
for i, value in enumerate(measurements):
    print("index =", i, "value =", value)
````

### `while`
A `while` loop repeats **while a condition remains true**. Because the condition is checked repeatedly, something inside the loop must eventually make the condition false—or the loop will run forever.


````python
count = 0
while count < 3:
    print("count =", count)
    count = count + 1
````

## 8. Explicit loops versus vectorized NumPy operations

For many numerical tasks, NumPy lets us replace a Python loop with an array expression. Both approaches below square every value.


````python
values = np.array([1.0, 2.0, 3.0, 4.0])

# Explicit Python loop
squared_loop = []
for value in values:
    squared_loop.append(value**2)

# Vectorized NumPy expression
squared_vectorized = values**2

print("loop result:      ", squared_loop)
print("vectorized result:", squared_vectorized)
````

The two approaches are conceptually equivalent here. For numerical work, the vectorized NumPy expression is usually shorter and, for large arrays, much faster. Explicit loops are still essential when each iteration requires complicated logic or when the task is not naturally an array operation.
